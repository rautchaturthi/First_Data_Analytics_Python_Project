### Step 01 : Add data into to do Analysis ###

```python
sales_data = [
    {"product": "Laptop", "quantity": 2, "price": 50000},
    {"product": "Mouse", "quantity": 10, "price": 500},
    {"product": "Keyboard", "quantity": 5, "price": 1500},
    {"product": "Laptop", "quantity": 1, "price": 50000},
    {"product": "Mouse", "quantity": 15, "price": 500},
    {"product": "Monitor", "quantity": 3, "price": 12000},
    {"product": "Keyboard", "quantity": 8, "price": 1500}
]

print("Total records:", len(sales_data))
```

### Step 02 : Analysis ###

que 01 : Total revenue generated

```python
total_revenue = 0

for sale in sales_data:
    revenue = sale["quantity"] * sale["price"]
    total_revenue += revenue

print("Total Revenue: ₹", total_revenue)
```

que 2 : Total number of products sold

```python
total_quantity = 0

for sale in sales_data:
    total_quantity += sale["quantity"]

print("Total Products Sold:", total_quantity)
```

Que 3 : Best-selling product

```python
product_sales = {}

for sale in sales_data:
    product = sale["product"]
    quantity = sale["quantity"]

    if product in product_sales:
        product_sales[product] += quantity
    else:
        product_sales[product] = quantity

best_selling = max(product_sales, key=product_sales.get)

print("Best-Selling Product:", best_selling)
print("Units Sold:", product_sales[best_selling])
```

Que 4: Product with the highest revenue

```python
product_revenue = {}

for sale in sales_data:
    product = sale["product"]
    revenue = sale["quantity"] * sale["price"]

    if product in product_revenue:
        product_revenue[product] += revenue
    else:
        product_revenue[product] = revenue

highest_revenue = max(product_revenue, key=product_revenue.get)

print("Highest Revenue Product:", highest_revenue)
print("Revenue: ₹", product_revenue[highest_revenue])
```

Que 5: Average sales per transaction

```python
total_revenue = 0

for sale in sales_data:
    total_revenue += sale["quantity"] * sale["price"]

total_transactions = len(sales_data)

average_sales = total_revenue / total_transactions

print("Average Sales per Transaction: ₹", round(average_sales, 2))
```
